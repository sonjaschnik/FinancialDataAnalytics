# Advice for working with Git

Git liest die Dateien direkt von der Festplatte. Du arbeitest ganz normal mit dem Windows Explorer cmd oder deinem Editor – und sagst Git Bash dann über cd einfach wo der Ordner ist.

### 1 Git Bash öffnen

Rechtsklick auf den gewünschten Ordner in meinen Dateien


### 2 Prüfen, ob du im richtigen Ordner bist

pwd	# zeigt deinen aktuellen Pfad
ls         # zeigt alle Dateien im Ordner


### 3 Create neues Repository oder bestehendes Github-Repository herunterladen

create:	git init

oder 

herunterladen:	git clone https://github.com/dein-nutzer/dein-repo.git
(dann in geklonten Ordner arbeiten)


### 4 Stage + Commit + Push





```python
# 1. Zeige an, was sich geändert hat
git status

# 2. Stage – einzelne Datei
git add meinedatei.txt

# ...oder ALLES auf einmal stagen
git add .

# 3. Commit mit Nachricht
git commit -m "Meine Änderung beschreiben"

# 4. Push zu GitHub
git push
```
